TASK 6

```python
# Check column names
print(internet.columns)
print(population.columns)
print(gdp.columns)
print(literacy.columns)


population.rename(columns={
    'Country': 'country',
    'Population': 'population'
}, inplace=True)
```

TASK 7

```python
gdp.info()

gdp['GDP'] = pd.to_numeric(
    gdp['GDP'], errors='coerce')
```

```python

literacy.info()
literacy['literacy_rate'] = pd.to_numeric(
    literacy['literacy_rate'], errors='coerce')
```

```python
internet.info()
internet['internet_users'] = pd.to_numeric(
    internet['internet_users'], errors='coerce')
```

```python
population.info()
population['population'] = pd.to_numeric(
    population['population'], errors='coerce'
)
```

TASK 8

```python
# Standardize column names before merging
gdp_renamed = gdp.rename(columns={'Country': 'country'})
literacy_renamed = literacy.rename(columns={'name': 'country'})

# Perform the outer joins
finaldf = pd.merge(population, gdp_renamed, on='country', how='outer')
finaldf = pd.merge(finaldf, internet, on='country', how='outer')
finaldf = pd.merge(finaldf, literacy_renamed, on='country', how='outer')

finaldf.head()
```

TASK 9

```python
finaldf['internet_penetration'] = ( finaldf['internet_users'] / finaldf['population']) * 100

finaldf['internet_penetration'] = finaldf[ 'internet_penetration'].round(2)
finaldf.head()
```

```python
finaldf['gdp_per_capita'] = (finaldf['GDP'] / finaldf['population'])

finaldf['gdp_per_capita'] = finaldf['gdp_per_capita'].round(2)
finaldf.head()
```

TASK 10

```python
import sqlite3

conn = sqlite3.connect('/content/country.db')

# Drop the duplicate uppercase 'GDP' column since we have the lowercase 'gdp' column
if 'GDP' in finaldf.columns and 'gdp' in finaldf.columns:
    df_to_save = finaldf.drop(columns=['GDP'])
else:
    df_to_save = finaldf

df_to_save.to_sql(
    'country_statistics',
    conn,
    if_exists='replace',
    index=False
)

print("Data successfully stored in database")
```

```python
#12 Find the average literacy rate.
query = """
SELECT AVG(literacy_rate) AS average_literacy_rate
FROM country_statistics
"""

pd.read_sql(query, conn)
```

```python
#13 Display all countries having literacy rate above 90%.
query = """
SELECT country, literacy_rate FROM country_statistics
WHERE literacy_rate > 90
"""

pd.read_sql(query, conn)
```

```python
#14 Find the Top 5 countries with the highest internet penetration.
query = """
SELECT country, internet_penetration FROM country_statistics
ORDER BY internet_penetration DESC
LIMIT 5
"""

pd.read_sql(query, conn)
```

```python
#15 Calculate the total population.
query = """
SELECT SUM(population) AS total_population
FROM country_statistics
"""

pd.read_sql(query, conn)
```

```python
#16 Find the country with the highest GDP per capita.
query = """
SELECT country, gdp_per_capita FROM country_statistics
ORDER BY gdp_per_capita DESC
LIMIT 1
"""

pd.read_sql(query, conn)
```
